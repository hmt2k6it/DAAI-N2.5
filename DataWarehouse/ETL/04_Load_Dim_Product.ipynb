# Tạo và Nạp dữ liệu cho bảng DIM_PRODUCT
Bảng này lấy từ file `products_silver.csv`. Đổi tên cột `price` thành `list_price` và tạo khóa chính `product_key`.

```python
import os
from pathlib import Path

import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

df_prod = pd.read_csv('../../SilverData/products_silver.csv')

df_prod['product_key'] = df_prod.index + 1
df_prod['product_id'] = df_prod['product_id'].astype(str)
df_prod['price'] = df_prod['price'].round(2)
df_prod['cogs'] = df_prod['cogs'].round(2)

cols = ['product_key', 'product_id', 'product_name', 'category', 'segment', 'size', 'color', 'price', 'cogs']
df_prod = df_prod[cols]

print(df_prod.head())
```

### Nạp dữ liệu (Load) vào Microsoft SQL Server


```python
from db_engine import get_engine

engine = get_engine()

df_prod.to_sql(
    name='dim_product',
    con=engine,
    if_exists='append',
    index=False,
    chunksize=2000,
)
print(f"Đã nạp thành công {len(df_prod):,} dòng vào DIM_PRODUCT trên SQL Server!")
```

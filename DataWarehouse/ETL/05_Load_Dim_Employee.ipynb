# Tạo và Nạp dữ liệu cho bảng DIM_EMPLOYEE
Bảng này lấy từ file `employees_silver.csv`, chỉ giữ lại thông tin cần thiết phục vụ báo cáo doanh số nhân viên: `sales_employee_id`, `sales_employee_name` và khóa chính `employee_key`.

```python
import pandas as pd
from sqlalchemy import create_engine
import getpass

# 1. Đọc dữ liệu nhân viên từ SilverData
df_emp = pd.read_csv('../../SilverData/employees_silver.csv')

# 2. Transform dữ liệu
# Tạo khóa chính employee_key
df_emp['employee_key'] = df_emp.index + 1

# Chọn đúng các cột trong schema SQL: employee_key, sales_employee_id, sales_employee_name
df_emp = df_emp[['employee_key', 'sales_employee_id', 'sales_employee_name']]

print("Dữ liệu DIM_EMPLOYEE mẫu:")
print(df_emp.head())
print(f"\nTổng số nhân viên: {len(df_emp)}")
```

### Nạp dữ liệu (Load) vào PostgreSQL

```python
# Yêu cầu nhập mật khẩu bảo mật
password = getpass.getpass("Nhập password cho tài khoản avnadmin: ")

# Kết nối Database vào datawarehouse
DB_URL = f"postgresql://avnadmin:{password}@itdocs-student-b775.f.aivencloud.com:11415/datawarehouse?sslmode=require"
engine = create_engine(DB_URL)

# Nạp vào bảng dim_employee
df_emp.to_sql('dim_employee', con=engine, if_exists='append', index=False)

print("\nĐã nạp dữ liệu vào bảng DIM_EMPLOYEE thành công!")
```

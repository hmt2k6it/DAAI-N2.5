# Tạo và Nạp dữ liệu cho bảng DIM_PROMOTION
Bảng này lấy từ file `promotions_silver.csv`, lọc các cột mô tả chương trình khuyến mãi: `promo_id`, `promo_name`, `promo_type` và gán khóa chính `promotion_key`.

```python
import pandas as pd
from sqlalchemy import create_engine
import getpass

# 1. Đọc dữ liệu khuyến mãi từ SilverData
df_promo = pd.read_csv('../../SilverData/promotions_silver.csv')

# 2. Transform dữ liệu
df_promo['promotion_key'] = df_promo.index + 1

# Sắp xếp đúng theo schema SQL: promotion_key, promo_id, promo_name, promo_type
cols = ['promotion_key', 'promo_id', 'promo_name', 'promo_type']
df_promo = df_promo[cols]

print("Dữ liệu DIM_PROMOTION mẫu:")
print(df_promo.head())
print(f"\nTổng số chương trình khuyến mãi: {len(df_promo)}")
```

### Nạp dữ liệu (Load) vào PostgreSQL

```python
# Yêu cầu nhập mật khẩu bảo mật
password = getpass.getpass("Nhập password cho tài khoản avnadmin: ")

# Kết nối Database vào datawarehouse
DB_URL = f"postgresql://avnadmin:{password}@itdocs-student-b775.f.aivencloud.com:11415/datawarehouse?sslmode=require"
engine = create_engine(DB_URL)

# Nạp vào bảng dim_promotion
df_promo.to_sql('dim_promotion', con=engine, if_exists='append', index=False)

print("\nĐã nạp dữ liệu vào bảng DIM_PROMOTION thành công!")
```

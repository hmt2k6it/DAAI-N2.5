```python
import pandas as pd
import numpy as np
df_products = pd.read_excel(r'C:\Users\Binh\OneDrive\Documents\IUH\AI\Project\DAAI-N2.5\Dictionary\products.xlsx')
```

```python
# 2. Xóa các dòng bị lặp hoàn toàn
df_products.drop_duplicates(inplace=True)
```

```python
# 3. Chuẩn hóa chuỗi văn bản (Xóa khoảng trắng thừa)
str_cols = ['product_name', 'category', 'segment', 'size', 'color']
for col in str_cols:
    if col in df_products.columns:
        df_products[col] = df_products[col].astype(str).str.strip()
```

```python
# 4. Kiểm tra kiểu dữ liệu và ép kiểu (chỉ thực hiện nếu cột tồn tại)
num_cols = ['product_id', 'price', 'cogs']
for col in num_cols:
    if col in df_products.columns:
        df_products[col] = pd.to_numeric(df_products[col], errors='coerce')
    else:
        print(f"Cảnh báo: Không tìm thấy cột '{col}' trong dataset!")
```

```python

required_cols = ['product_id', 'product_name', 'category', 'segment', 'size', 'color', 'price', 'cogs']
existing_cols = [col for col in required_cols if col in df_products.columns]

# Lọc bỏ dữ liệu khuyết ở các cột hiện có
df_products.dropna(subset=existing_cols, inplace=True)
```

```python
# 8. Xuất file kết quả tầng Silver
df_products.to_csv('products_silver.csv', index=False)
print("Hoàn tất xử lý Silver Data cho Products!")
```

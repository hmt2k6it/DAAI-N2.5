```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../customers.csv')
df.head()
```

```python
# Xem kiểu dữ liệu và có giá trị null không
df.info()
```

```python
# Kiểm tra thử có id trùng lặp không
df['customer_id'].is_unique
```

```python
# Xem thử có bao nhiều giá trị trong gender, có gì trị nào không hợp lý không 
df['gender'].value_counts()
```

```python
# Xem thử có bao nhiều giá trị trong age_group
df['age_group'].value_counts()
```

```python
# Xem thử có bao nhiều giá trị trong acquisition_channel
df['acquisition_channel'].value_counts()
```

```python
# Chuyển cột ngày sang datetime (errors='coerce': ngày sai sẽ thành NaT để dễ phát hiện)
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
print('Số ngày không đọc được:', df['signup_date'].isna().sum())
df['gender'] = df['gender'].astype('category')
df['age_group'] = df['age_group'].astype('category')
df['acquisition_channel'] = df['acquisition_channel'].astype('category')
df.info()
```

```python
# Đối chiếu master data: zip và city phải tồn tại trong bảng geography
geo = pd.read_csv('../geography.csv')
print('Số zip không tồn tại trong geography:', (~df['zip'].isin(geo['zip'])).sum())
print('Số city không tồn tại trong geography:', (~df['city'].isin(geo['city'])).sum())
```

```python
# Tổng kết trước khi xuất: không còn trùng, không còn null
print('Tổng số dòng     :', len(df))
print('Tổng số dòng trùng customer_id:', df.duplicated(subset=['customer_id']).sum())
print('Tổng số null     :', df.isna().sum().sum())
```

```python
# Xuất dữ liệu đã được làm sạch ra file CSV mới
df.to_csv('../SilverData/customers.csv', index=False)
```

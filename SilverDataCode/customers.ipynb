```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../customers.csv')
df.head()
```

```python
# Xem kiểu dữ liệu và có giá trị null không
df.info()
```

```python
# Kiểm tra thử có id trùng lặp không
df['customer_id'].is_unique
```

```python
# Xem thử có bao nhiều giá trị trong gender, có gì trị nào không hợp lý không 
df['gender'].value_counts()
```

```python
# Xem thử có bao nhiều giá trị trong age_group
df['age_group'].value_counts()
```

```python
# Xem thử có bao nhiều giá trị trong acquisition_channel
df['acquisition_channel'].value_counts()
```

```python
# Chuyển đổi các cột dữ liệu sang định dạng phù hợp
df['signup_date'] = pd.to_datetime(df['signup_date'])
df['gender'] = df['gender'].astype('category')
df['age_group'] = df['age_group'].astype('category')
df['acquisition_channel'] = df['acquisition_channel'].astype('category')
df.info()
```

```python
# Xuất dữ liệu đã được làm sạch ra file CSV mới
df.to_csv('../SilverData/customers.csv', index=False)
```

```python
import pandas as pd
import numpy as np

df = pd.read_json("../epd.json")
df.head()
```

```python
df.info()
```

```python
# Ta chuyển product_id sang kiểu int64 để dễ xử lý hơn
df["product_id"] = pd.to_numeric(df["product_id"], errors="coerce").astype("Int64")
# Kiểm tra các giá trị product_id bị NaN
display(df[df["product_id"].isna()])
```

```python
# Vì product_id bị nan, mà ta thấy product_id có thể là index + 1
# nên ta sẽ thử giá trị 32 cho dòng 31 và 288 cho dòng 287 đã có chưa
display(df[df["product_id"].isin([32, 288])])
```

```python
# Vì chưa có nên ta sẽ gán giá trị product_id = index + 1 cho dòng 31 product_id bị null
df.loc[31, 'product_id'] = 32
df.loc[287, 'product_id'] = 288
```

```python
# Kiểm tra các dòng trùng ID
display(df[df["product_id"].duplicated(keep=False)])
```

```python
# Có 2 dòng bị trung product_id nhưng nó lại khác các giá trị khác nên ta cũng sẽ
# giữ và gán product_id của index 403 là 404
# Xem thử product_id 404 đã có chưa ?
display(df[df['product_id'] == 404])
```

```python
# Vì không có nên ta sẽ gán giá trị product_id = index + 1 cho dòng 403 product_id bị trùng
df.loc[403, 'product_id'] = 404
```

```python
# Xem các giá trị danh mục
print(df["category"].value_counts(dropna=False))
```

```python
# Xem thử 2 giá trị N/A và Unknown_999 có phải là NaN không
display(df[df["category"].isnull() | df["category"].isin(["N/A", "Unknown_999"])])
```

```python
# Chuyển danh mục không xác định thành nan
df["category"] = df["category"].replace(["N/A", "Unknown_999"], np.nan)
# Bỏ dòng thiếu danh mục
df = df.dropna(subset=["category"])
```

```python
# Xem các giá trị segment
print(df["segment"].value_counts(dropna=False))
```

```python
# Xem các giá trị size
print(df["size"].value_counts(dropna=False))
```

```python
# Xem các giá trị color
print(df["color"].value_counts(dropna=False))
```

```python
# Xem thử có giá trị price nào bị NaN không ?
# Vì price đang là object nên ta sẽ sử chuyển sang float để kiểm tra
df["price"] = pd.to_numeric(df["price"], errors="coerce")
display(df[df["price"].isna()])
```

```python
# Chuyển giá không đọc được thành NaN và đổi sang số
df["price"] = pd.to_numeric(df["price"], errors="coerce")
```

```python
# Quyết định bỏ dòng có giá trị price bị NaN
df = df.dropna(subset=["price"])
```

```python
# Check thử có giá vốn và giá bán bị âm không
display(df[df["price"] < 0])
display(df[df["cogs"] < 0])
```

```python
# Giả định giá âm do nhập nhầm dấu, đổi -150 thành 150
df.loc[72, "price"] = 150

# Kiểm tra lại giá âm
display(df[df["price"] < 0])
```

```python
# Chuyển kiểu dữ liệu category cho các cột danh mục
df["category"] = df["category"].astype("category")
df["segment"] = df["segment"].astype("category")
df["size"] = df["size"].astype("category")
df["color"] = df["color"].astype("category")
```

```python
# Reset index rồi kiểm tra kết quả cuối
df = df.reset_index(drop=True)
df.info()
```

```python
# Tổng kết trước khi xuất: không còn trùng, không còn null
print('Tổng số dòng     :', len(df))
print('Tổng số dòng trùng product_id:', df.duplicated(subset=['product_id']).sum())
print('Tổng số null     :', df.isna().sum().sum())
```

```python
# Xuất CSV không kèm cột index
df.to_csv("../SilverData/epd.csv", index=False)
```

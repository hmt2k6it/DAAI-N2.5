```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../geography.csv')
df.head()
```

```python
df.info()
```

```python
#Kiểm tra thử 
display(df["zip"].is_unique)
display(df["zip"].isna().sum())
```

```python
# Kiểm tra thử city có bị trống hay không ?
display(df[(df["city"] == "") | (df["city"].isnull())])
```

```python
# Kiểm tra thử region có bị trống hay không ?
display(df[(df["region"] == "") | (df["region"].isnull())])
```

```python
# Kiểm tra thử district có bị trống hay không ?
display(df[(df["district"] == "") | (df["district"].isnull())])
```

```python
# zip là khóa chính của bảng địa lý; bộ ba (city, region, district) lặp lại là bình thường (nhiều zip cùng một quận)
print('Trùng zip   :', df.duplicated(subset=['zip']).sum())
print('Trùng bộ ba :', df.duplicated(subset=['city', 'region', 'district']).sum(), '(bình thường)')

# Chuyển cột phân loại sang category
for col in ['city', 'region', 'district']:
    df[col] = df[col].astype('category')


```

```python
# Tổng kết trước khi xuất
print('Tổng số dòng     :', len(df))
print('Tổng số zip trùng:', df.duplicated(subset=['zip']).sum())
print('Tổng số null     :', df.isna().sum().sum())
```

```python
df.to_csv("../SilverData/geography.csv", index=False)
```

```python
import pandas as pd
```

```python
df = pd.read_csv("../payments.csv")
df.head()
```

```python
# Xem kiểu dữ liệu và có giá trị null không
df.info()
```

```python
# Thống kê mô tả các cột số
df.describe()
```

```python
# Kiểm tra số lượng null trong các cột
df.isna().sum()
```

```python
# Kiểm tra thử có order_id trùng lặp không
print("order_id is unique     :", df["order_id"].is_unique)
print("Số dòng trùng order_id :", df.duplicated(subset=["order_id"]).sum())
```

```python
# Xem thử có bao nhiêu giá trị trong payment_method
df["payment_method"].value_counts(dropna=False)
```

```python
# Xem thử phân bố số kỳ trả góp (installments)
df["installments"].value_counts(dropna=False).sort_index()
```

```python
# Kiểm tra các giá trị bất thường (payment_value <= 0 hoặc COD trả góp > 1)
print("payment_value <= 0:", (df["payment_value"] <= 0).sum())
print("COD có trả góp > 1 :", ((df["payment_method"] == "cod") & (df["installments"] > 1)).sum())
```

```python
# Chuẩn hóa khoảng trắng cho cột payment_method
df["payment_method"] = df["payment_method"].astype(str).str.strip()
```

```python
# Ép kiểu dữ liệu chuẩn (Schema mapping)
df["order_id"] = df["order_id"].astype("int32")
df["installments"] = df["installments"].astype("int8")
df["payment_method"] = df["payment_method"].astype("category")
df["payment_value"] = df["payment_value"].astype("float64")

df.info()
```

```python
# Sắp xếp theo order_id và reset index
df = df.sort_values("order_id").reset_index(drop=True)

# Tổng kết trước khi xuất: không còn trùng, không còn null
print("Tổng số dòng     :", len(df))
print("Tổng số dòng trùng order_id:", df.duplicated(subset=["order_id"]).sum())
print("Tổng số null     :", df.isna().sum().sum())
```

```python
# Xuất dữ liệu đã được làm sạch ra file CSV mới
df.to_csv("../SilverData/payments.csv", index=False)
print("Exported payments.csv successfully!")
```

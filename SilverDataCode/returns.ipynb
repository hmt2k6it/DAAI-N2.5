```python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu thô từ file Excel
df_returns = pd.read_excel(r'C:\Users\Binh\OneDrive\Documents\IUH\AI\Project\DAAI-N2.5\Dictionary\returns.xlsx')
```

```python
# 2. Xóa trùng lặp
df_returns.drop_duplicates(inplace=True)
```

```python
# 3. Chuẩn hóa chuỗi văn bản
str_cols = ['return_id', 'return_reason']
for col in str_cols:
    if col in df_returns.columns:
        df_returns[col] = df_returns[col].astype(str).str.strip()
```

```python
# 4. Ép kiểu dữ liệu số & ngày tháng
num_cols = ['order_id', 'product_id', 'return_quantity', 'refund_amount']
for col in num_cols:
    if col in df_returns.columns:
        df_returns[col] = pd.to_numeric(df_returns[col], errors='coerce')

if 'return_date' in df_returns.columns:
    df_returns['return_date'] = pd.to_datetime(df_returns['return_date'], errors='coerce')
```

```python
# 6. Lọc ràng buộc logic
if 'return_quantity' in df_returns.columns and 'refund_amount' in df_returns.columns:
    df_returns = df_returns[(df_returns['return_quantity'] > 0) & (df_returns['refund_amount'] >= 0)]
```

```python
# 5. Loại bỏ dữ liệu missing
required_cols = ['return_id', 'order_id', 'product_id', 'return_date', 'return_quantity', 'refund_amount']
existing_cols = [col for col in required_cols if col in df_returns.columns]
if existing_cols:
    df_returns.dropna(subset=existing_cols, inplace=True)
```

```python
# 7. Xuất file Silver Data
df_returns.to_csv('returns_silver.csv', index=False)
print("Hoàn tất xử lý Silver Data cho Returns!")
```

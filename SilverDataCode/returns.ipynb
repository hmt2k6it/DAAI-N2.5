```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../returns.csv')
df.head()
```

```python
# Xem kiểu dữ liệu và kiểm tra null
df.info()
```

```python
# Theo Dictionary, return_id là PK => kiểm tra trùng lặp
print('return_id là unique:', df['return_id'].is_unique)
print('Số dòng trùng toàn bộ:', df.duplicated().sum())
```

```python
# Chuyển cột return_date sang datetime
df['return_date'] = pd.to_datetime(df['return_date'], errors='coerce')
print('Số ngày không đọc được:', df['return_date'].isna().sum())
```

```python
# Xem các lý do trả hàng
df['return_reason'].value_counts()
```

```python
# Đối chiếu FK: order_id phải tồn tại trong bảng orders
orders = pd.read_csv('../orders_enriched.csv', usecols=['order_id'])
print('Số order_id không tồn tại trong orders:', (~df['order_id'].isin(orders['order_id'])).sum())
```

```python
# Đối chiếu FK: product_id phải tồn tại trong bảng products
products = pd.read_csv('../products.csv', usecols=['product_id'])
print('Số product_id không tồn tại trong products:', (~df['product_id'].isin(products['product_id'])).sum())
```

```python
# Chuyển kiểu dữ liệu
df['return_reason'] = df['return_reason'].astype('category')
df.info()
```

```python
# Tổng kết trước khi xuất
print('Tổng số dòng     :', len(df))
print('Tổng số dòng trùng return_id:', df.duplicated(subset=['return_id']).sum())
print('Tổng số null     :', df.isna().sum().sum())
```

```python
# Xuất dữ liệu đã được làm sạch ra file CSV mới
df.to_csv('../SilverData/returns.csv', index=False)
```

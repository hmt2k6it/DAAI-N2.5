```python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu thô từ file Excel
df_reviews = pd.read_excel(r'C:\Users\Binh\OneDrive\Documents\IUH\AI\Project\DAAI-N2.5\Dictionary\reviews.xlsx')
```

```python
# 2. Xóa các bản ghi trùng lặp
df_reviews.drop_duplicates(inplace=True)
```

```python
# 3. Chuẩn hóa văn bản
str_cols = ['review_id', 'review_title']
for col in str_cols:
    if col in df_reviews.columns:
        df_reviews[col] = df_reviews[col].astype(str).str.strip()
```

```python
# 4. Ép kiểu dữ liệu
num_cols = ['order_id', 'product_id', 'customer_id', 'rating']
for col in num_cols:
    if col in df_reviews.columns:
        df_reviews[col] = pd.to_numeric(df_reviews[col], errors='coerce')

if 'review_date' in df_reviews.columns:
    df_reviews['review_date'] = pd.to_datetime(df_reviews['review_date'], errors='coerce')
```

```python
# 5. Lọc bỏ các dòng bị Null
required_cols = ['review_id', 'order_id', 'product_id', 'customer_id', 'rating', 'review_date']
existing_cols = [col for col in required_cols if col in df_reviews.columns]
if existing_cols:
    df_reviews.dropna(subset=existing_cols, inplace=True)
```

```python
# 6. Ràng buộc Rating phải nằm trong khoảng từ 1 đến 5
if 'rating' in df_reviews.columns:
    df_reviews = df_reviews[(df_reviews['rating'] >= 1) & (df_reviews['rating'] <= 5)]
```

```python
# 7. Xuất file Silver Data
df_reviews.to_csv('reviews_silver.csv', index=False)
print("Hoàn tất xử lý Silver Data cho Reviews!")
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../reviews.csv')
df.head()
```

```python
# Xem kiểu dữ liệu và kiểm tra null
df.info()
```

```python
# Theo Dictionary, review_id là PK => kiểm tra trùng lặp
print('review_id là unique:', df['review_id'].is_unique)
```

```python
# Chuyển cột review_date sang datetime
df['review_date'] = pd.to_datetime(df['review_date'], errors='coerce')
print('Số ngày không đọc được:', df['review_date'].isna().sum())
```

```python
# Xem phân bố rating (1-5)
df['rating'].value_counts().sort_index()
```

```python
# Xem các review_title
df['review_title'].value_counts()
```

```python
# Đối chiếu FK: order_id phải tồn tại trong bảng orders
orders = pd.read_csv('../orders_enriched.csv', usecols=['order_id'])
print('Số order_id không tồn tại trong orders:', (~df['order_id'].isin(orders['order_id'])).sum())
```

```python
# Đối chiếu FK: product_id phải tồn tại trong bảng products
products = pd.read_csv('../products.csv', usecols=['product_id'])
print('Số product_id không tồn tại trong products:', (~df['product_id'].isin(products['product_id'])).sum())
```

```python
# Đối chiếu FK: customer_id phải tồn tại trong bảng customers
customers = pd.read_csv('../customers.csv', usecols=['customer_id'])
print('Số customer_id không tồn tại trong customers:', (~df['customer_id'].isin(customers['customer_id'])).sum())
```

```python
# Chuyển kiểu dữ liệu
df['review_title'] = df['review_title'].astype('category')
df.info()
```

```python
# Tổng kết trước khi xuất
print('Tổng số dòng     :', len(df))
print('Tổng số dòng trùng review_id:', df.duplicated(subset=['review_id']).sum())
print('Tổng số null     :', df.isna().sum().sum())
```

```python
# Xuất dữ liệu đã được làm sạch ra file CSV mới
df.to_csv('../SilverData/reviews_silver.csv', index=False)
```
